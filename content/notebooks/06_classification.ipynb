# Classification

Classification means predicting a **category**, such as apple or orange, from measurements.

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

fruit = pd.read_csv("data/simple_classification.csv")
fruit.head()
```

## Look first

```python
for name, group in fruit.groupby("fruit"):
    plt.scatter(group["weight_g"], group["diameter_cm"], label=name)
plt.xlabel("Weight (g)")
plt.ylabel("Diameter (cm)")
plt.legend()
plt.show()
```

## Train and test a K-nearest neighbours model

KNN looks at the closest known examples and takes a vote.

```python
X = fruit[["weight_g", "diameter_cm"]]
y = fruit["fruit"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
pred = knn.predict(X_test)
print("Accuracy:", accuracy_score(y_test, pred))
print(confusion_matrix(y_test, pred))
```

## Try a decision tree

```python
tree = DecisionTreeClassifier(max_depth=3, random_state=0)
tree.fit(X_train, y_train)
print("Tree accuracy:", accuracy_score(y_test, tree.predict(X_test)))
```

## Overfitting

A model that is too complicated can memorise the training data and then do worse on new data. Compare the scores below.

```python
for depth in [1, 2, 3, 5, 10]:
    t = DecisionTreeClassifier(max_depth=depth, random_state=0).fit(X_train, y_train)
    print(depth, "train:", round(t.score(X_train, y_train), 2), "test:", round(t.score(X_test, y_test), 2))
```
