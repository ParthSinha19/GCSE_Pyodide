## Variables and numbers

```python
name = "Sam"
age = 14
height = 1.62

print(name, age, height)
print(age + 1)
print(age * 2, age ** 2, 17 // 5, 17 % 5)
```

## Strings

```python
message = "hello"
print(message.upper())
print(len(message))
print(message + " world")
print(f"{name} is {age} years old")
```

## Lists

```python
scores = [12, 18, 15, 20]
scores.append(17)
print(scores)
print(scores[0], scores[-1])
print(sum(scores) / len(scores))
```

## Loops

```python
for s in scores:
    print("Score:", s)

for i in range(3):
    print("Count", i)
```

## Conditions

```python
for s in scores:
    if s >= 18:
        print(s, "great")
    elif s >= 15:
        print(s, "good")
    else:
        print(s, "keep practising")
```

## Functions

```python
def average(numbers):
    return sum(numbers) / len(numbers)

print(average(scores))
```
