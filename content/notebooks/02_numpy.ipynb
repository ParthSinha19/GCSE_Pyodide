# NumPy

NumPy lets us do maths on whole lists of numbers at once.

```python
import numpy as np

a = np.array([3, 5, 8, 2, 9])
print(a)
print(a + 10)
print(a * a)
```

## Statistics

```python
print("mean:", a.mean())
print("median:", np.median(a))
print("max:", a.max(), "min:", a.min())
print("std:", a.std())
```

## Indexing and slicing

```python
print(a[0], a[-1])
print(a[1:4])
print(a[a > 4])
```

## Making arrays

```python
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
print(np.zeros(4))
print(np.ones((2, 3)))
```

## Random numbers

The `seed` makes the "random" numbers repeat, so everyone gets the same answer.

```python
rng = np.random.default_rng(seed=1)
print(rng.integers(1, 7, size=10))   # ten dice rolls
print(rng.normal(0, 1, size=5))
```

## 2D arrays and a little linear algebra

```python
m = np.array([[1, 2], [3, 4]])
print(m)
print(m.T)
print(m @ m)
```
