# Matplotlib

Graphs help us see patterns in numbers.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
```

## Line graph

```python
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Sine and cosine")
plt.legend()
plt.grid(True)
plt.show()
```

## Scatter plot

```python
rng = np.random.default_rng(2)
hours = rng.uniform(0, 10, 30)
score = 40 + 5 * hours + rng.normal(0, 6, 30)

plt.scatter(hours, score)
plt.xlabel("Hours revised")
plt.ylabel("Test score")
plt.title("Does revision help?")
plt.show()
```

## Bar chart

```python
subjects = ["Maths", "English", "Science", "Art"]
marks = [72, 65, 80, 58]
plt.bar(subjects, marks, color="steelblue")
plt.ylabel("Average mark")
plt.title("Marks by subject")
plt.show()
```

## Histogram

```python
values = rng.normal(50, 10, 200)
plt.hist(values, bins=10, edgecolor="black")
plt.xlabel("Value")
plt.ylabel("How many")
plt.show()
```

## Subplots

```python
fig, axes = plt.subplots(1, 2, figsize=(9, 3))
axes[0].plot(x, x**2)
axes[0].set_title("x squared")
axes[1].plot(x, np.sqrt(x))
axes[1].set_title("square root of x")
plt.tight_layout()
plt.show()
```
