# Working with data

We will use **pandas** to read a CSV file (a table) and explore it. The file `data/students.csv` has made-up revision and test-score data.

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/students.csv")
df.head()
```

## Look at the data

```python
print(df.shape)
print(df.columns.tolist())
df.describe()
```

## Pick columns and filter rows

```python
print(df["test_score"].mean())
high = df[df["test_score"] >= 80]
print(len(high), "students scored 80 or more")
```

## Group and compare

```python
df.groupby("year_group")["test_score"].mean()
```

## Look for patterns

```python
plt.scatter(df["hours_revised"], df["test_score"])
plt.xlabel("Hours revised")
plt.ylabel("Test score")
plt.show()

print(df[["hours_revised", "hours_sleep", "test_score"]].corr())
```
