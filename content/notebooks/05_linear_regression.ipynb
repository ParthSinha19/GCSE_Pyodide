# Linear regression

Can we predict a house price from its size? Linear regression draws the best straight line through the data.

- **Features** are the inputs (here, size).
- The **label** is what we want to predict (here, price).

```python
%matplotlib inline
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

houses = pd.read_csv("data/house_prices.csv")
houses.head()
```

## Look first

```python
plt.scatter(houses["size_m2"], houses["price_thousands_gbp"])
plt.xlabel("Size (m²)")
plt.ylabel("Price (£1000s)")
plt.show()
```

## Split into training and testing data

We train on one part and test on data the model has never seen.

```python
X = houses[["size_m2"]]
y = houses["price_thousands_gbp"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
print(len(X_train), "training rows,", len(X_test), "testing rows")
```

## Train the model

```python
model = LinearRegression()
model.fit(X_train, y_train)
print("Each extra m² adds about £", round(model.coef_[0], 2), "thousand")
```

## Test the model

```python
predictions = model.predict(X_test)
print("Average error: £", round(mean_absolute_error(y_test, predictions), 1), "thousand")

plt.scatter(X_test, y_test, label="real")
plt.plot(X_test.sort_values("size_m2"), model.predict(X_test.sort_values("size_m2")), color="red", label="model")
plt.xlabel("Size (m²)")
plt.ylabel("Price (£1000s)")
plt.legend()
plt.show()
```
